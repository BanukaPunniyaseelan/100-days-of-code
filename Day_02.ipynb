```python
print("Welcome to the Tip Calculator!")
print("-----------------------------")

bill = input("Enter the bill amount:")
tip_percent = input("Enter tip percentage:")
people = input("How many people are splitting the bill?")

bill = float(bill)
tip_percent = float(tip_percent)
people = int(people)

tip_amount = bill * (tip_percent / 100)
total_amount = bill + tip_amount
amount_per_person = total_amount / people

print("\n----- Bill Summary -----")
print(f"Bill amount: {bill:.2f}")
print(f"Tip percentage: {tip_percent:.0f}%")
print(f"Tip amount: {tip_amount:.2f}")
print(f"Total amount: {total_amount:.2f}")
print(f"Each person pays: {amount_per_person:.2f}")
```
