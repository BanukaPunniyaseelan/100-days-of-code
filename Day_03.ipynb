```python
print("Welcome to the Name Formatter!")
print("-----------------------------")

first_name = input("Enter your first name:")
last_name = input("Enter your last name:")
favourite_word = input("Enter your favourite word:")

first_name = first_name.strip()
last_name = last_name.strip()
favourite_word = favourite_word.strip()

full_name = first_name + " " + last_name
upper_name = full_name.upper()
lower_name = full_name.lower()
title_name = full_name.title()

word_length = len(favourite_word)

print("\n----- Name Formatter Results -----")
print(f"Full name: {full_name}")
print(f"Uppercase name: {upper_name}")
print(f"Lowercase name: {lower_name}")
print(f"Title case name: {title_name}")
print(f"Your favourite word is: {favourite_word}")
print(f"It has {word_length} letters.")

sentence = f"{title_name} loves the word {favourite_word.upper()}."
print(sentence)
```
