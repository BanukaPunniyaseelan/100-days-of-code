```python
print("Welcome to the Contact Book!")
print("----------------------------")

contacts = {}

contacts["Banu"] = "0770450318"
contacts["Brindha"] = "0751198078"
contacts["Pavithran"] = "0784423108"

print('\n----- All Contacts -----')

for name in contacts:
  print(f"Name: {name}")
  print(f"Phone: {contacts[name]}")
  print("------------------------")

search_name = input("\nEnter a name to search: ")
if search_name in contacts:
  print(f"Phone number for {search_name}: {contacts[search_name]}")
else:
  print(f"No contact found for {search_name}.")

new_name = input("\nEnter a new contact name: ")
new_number = input("\nEnter a new phone number: ")

contacts[new_name] = new_number

print(f"\n{new_name} was added successfully.")

update_name = input("\nEnter a contact name to update: ")
if update_name in contacts:
  new_number = input("Enter the new phone number: ")
  contacts[update_name] = new_number
  print(f"{update_name}'s phone number was updated.")
else:
  print(f"No contact found for {update_name}.")

delete_name = input("\nEnter a contact name to delete: ")
if delete_name in contacts:
  del contacts[delete_name]
  print(f"{delete_name} was deleted.")
else:
  print(f"Np contact found for {delete_name}.")

print("\n----- Final Contact Book -----")

for name, phone in contacts.items():
  print(f"Name: {name}")
  print(f"Phone: {phone}")
  print("---------------------------")
```
