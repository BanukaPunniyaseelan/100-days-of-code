```python
print("Welcome to the Password Strength Checker!")
print("-----------------------------------------")

password = input("Enter a password: ")

length = len(password)

has_number = False

for character in password:
  if character.isdigit():
    has_number = True

has_uppercase = False

for character in password:
  if character.isupper():
    has_uppercase = True

has_lowercase = False

for character in password:
  if character.islower():
    has_lowercase = True

print("\n----- Password Analysis -----")
print(f"Password length: {length}")
print(f"Contains a number: {has_number}")
print(f"Contains an uppercase letter: {has_uppercase}")
print(f"Contains a lowercase letter: {has_lowercase}")

if length >= 8 and has_number and has_uppercase and has_lowercase:
  print("Password strength: Strong")
elif length >= 6 and (has_number or has_uppercase):
  print("Password strength: Medium")
else:
  print("Password strength: Weak")
```
