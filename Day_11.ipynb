```python
print("Wlcome to the Palindrome Checker!")
print("---------------------------------")

text = input("Enter a word: ")
clean_text = text.replace(" ", "").lower()
reversed_text = clean_text[::-1]

print("\n----- Palindrome Analysis -----")
print(f"Orginal text: {text}")
print(f"Cleaned text: {clean_text}")
print(f"Reversed text: {reversed_text}")

if clean_text == reversed_text:
  print(f"\n'{text}' is a palindrome.")
else:
  print(f"\n'{text}' is not a palindrome.")

letter_count = len(clean_text)
print(f"Total letters (without spaces): {letter_count}")
```
