```python
print("Welcome to the Multiplication Table Generator!")
print("----------------------------------------------")

number = input("Enter a number: ")
number = int(number)

print(f"\nMultiplication table for {number}:")

for i in range(0, 13):
  result = number * i
  print(f"{number} x {i} = {result}")

print("\nTable completed.")
```
