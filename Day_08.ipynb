```python
print("Welcome to the Favourite Movies List!")
print("-------------------------------------")

movies = []

for i in range(1, 7):
  movie = input(f"Enter movie number {i}: ")
  movies.append(movie)

print("\n----- Your Favourite Movies -----")

for movie in movies:
  print(f"- {movie}")

print("\n----- List Information -----")
print(f"Total movies: {len(movies)}")
print(f"First movie: {movies[0]}")
print(f"Last movie: {movies[-1]}")

movies.sort()

print(f"\n----- Movies in Alphabetical Order ------")
for movie in movies:
  print(f"- {movie}")

movie_to_remove = input("\nEnter a movie you want to remove: ")

if movie_to_remove in movies:
  movies.remove(movie_to_remove)
  print(f"\n'{movie_to_remove}' was removed.")
else:
  print(f"\n' {movie_to_remove}' is not in your list.")

print("\n----- Final Movie List -----")
for movie in movies:
  print(f"- {movie}")
```
